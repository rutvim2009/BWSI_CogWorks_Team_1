```python
import numpy as np
import random
import soundfile as sf
from IPython.display import Audio
# Take audio file in, return an array of "spliced" audio files 

def get_segment(wav_path, start_sec, duration_sec = 10):
    with sf.SoundFile(wav_path) as f: # with is good practice for sf ig
        samplerate = f.samplerate # sets samplerate to the samplerate(most like 44100)
        f.seek(int(start_sec * samplerate)) # .seek lets us pick where to start before reading, as the default is 0 only allowing reading from the start
        segment = f.read(int(duration_sec * samplerate)) # sets the segment to a portion of the audio
    return segment

def AudioSplicer(wav_path, duration_sec = 10, quantity = 4):
    audio_list = []
    for i in range(quantity):
        info = sf.info(wav_path)
        start_sec = random.random() * (info.duration - duration_sec)
        audio_list.append(get_segment(wav_path, start_sec, duration_sec))
    return audio_list
```
