```python
import pickle #import to save and load files

def get_song(self, song_id):#finds a specific song
    if song_id not in self.songs:
        print("ERROR: Song not found.")
        return None
    return self.songs.get(song_id)

def get_all_songs(self): #all songs are returned
    return self.songs

def num_songs(self): #total number of songs within the dictionary
    return len(self.songs)

def contains(self, new_title, new_artist): #checks if song is in database
    target = (new_title.lower(), new_artist.lower())
    for song in self.songs.values():
        stored = (song["title"].lower(), song["artist"].lower())
        if stored == target:
            return True
    return False

def delete_song(self, new_song_id): #deletes a song from the dictionary when ran
    if new_song_id not in self.songs:
        return self.songs
    self.songs.pop(new_song_id)
    return self.songs

#save/load
def save(self, path): #save the file in the system
    with open(path, mode="wb") as f:
        pickle.dump(self, f)

@classmethod #makes it easier to call, less work 
def load(cls, path): #loads the file 
    with open(path, "rb") as f:
        return pickle.load(f)
```
