```python
class SongMetadata:
    def __init__(self):
        self.songs = {}

    def add_song(self, title, artist, file_path):
        for song_id, song in self.songs.items():
            if (
                song["title"].lower() == title.lower()
                and song["artist"].lower() == artist.lower()
            ):
                return song_id

        song_id = str(uuid.uuid4())

        self.songs[song_id] = {
            "title": title,
            "artist": artist,
            "file_path": file_path
        }

        return song_id

    def get_song(self, song_id):
        return self.songs.get(song_id)

    def get_all_songs(self):
        return self.songs

    #added the inspections within the list here [Aahana]
    def load_song(self):
        return pickle.load(self.song_id)
            
    def num_songs(self): #total number of songs within the dictionary
        return len(self.songs)

    def contains(self, new_title, new_artist):
        target = (new_title.lower(), new_artist.lower())
        for song in self.songs.values():
            stored = (song["title"], song["artist"])
            if stored == target:
                return True
        return False

    def delete_song(self, new_song_id): #deletes a song from the dictionary when ran
        if new_song_id not in self.songs:
            return self.songs
        self.songs.pop(new_song_id)
```

```python
## Saving & loading — Shriyans
import pickle
import uuid


def save_database(fingerprint_db, metadata, filename):
    everything = {"fingerprint_db": fingerprint_db, "metadata": metadata}
    with open(filename, mode="wb") as opened_file:
        pickle.dump(everything, opened_file)


def load_database(filename):
    with open(filename, mode="rb") as opened_file:
        everything = pickle.load(opened_file)
    return everything["fingerprint_db"], everything["metadata"]
```

```python
## Saving & loading — Shriyans
test_db = {(100, 200, 5): [("song1", 0.5)], (150, 250, 3): [("song1", 1.2)]}
test_metadata = SongMetadata()
test_metadata.add_song("title1", "artist1", "Music/song1.mp3")

save_database(test_db, test_metadata, "test_db.pkl")
loaded_db, loaded_metadata = load_database("test_db.pkl")

print(loaded_db == test_db)
print(loaded_metadata.get_all_songs() == test_metadata.get_all_songs())
```
