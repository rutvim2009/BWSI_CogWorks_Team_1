```python
class SongMetadata:
    def __init__(self):
        self.songs = {}

    def add_song(self, title, artist, file_path):
        for song_id, song in self.songs.items():
            if (
                song["title"].lower() == title.lower()
                and song["artist"].lower() == artist.lower()
            ):
                return song_id

        song_id = str(uuid.uuid4())

        self.songs[song_id] = {
            "title": title,
            "artist": artist,
            "file_path": file_path
        }

        return song_id

    def get_song(self, song_id):
        return self.songs.get(song_id)

    def get_all_songs(self):
        return self.songs

    #added the inspections within the list here [Aahana]
    def num_songs(self): #total number of songs within the dictionary
        return len(self.songs)

    def contains(self, new_title, new_artist):
        target = (new_title.lower(), new_artist.lower())
        for song in self.songs.values():
            stored = (song["title"].strip().lower(), song["artist"].strip().lower())
            if stored == target:
                return True
        return False

    def delete_song(self, new_song_id): #deletes a song from the dictionary when ran
        if new_song_id not in self.songs:
            return self.songs
        self.songs.pop(new_song_id)
            
            
        
```
