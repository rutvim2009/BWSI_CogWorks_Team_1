```python
import uuid
import json
import os


class SongMetadata:
    def __init__(self):
        self.songs = {}

    def add_song(self, title, artist, file_path, recommendations=None):
        for song_id, song in self.songs.items():
            if (
                song["title"].lower() == title.lower()
                and song["artist"].lower() == artist.lower()
            ):
                return song_id

        song_id = str(uuid.uuid4())

        self.songs[song_id] = {
            "title": title,
            "artist": artist,
            "file_path": file_path,
            "recommendations": recommendations if recommendations else []
        }

        return song_id

    def get_song(self, song_id):
        return self.songs.get(song_id)

    def get_all_songs(self):
        return self.songs

    def get_recommendations(self, song_id):
        song = self.get_song(song_id)

        if song:
            return song["recommendations"]

        return []

    def add_recommendation(self, song_id, recommendation):
        if song_id in self.songs:
            self.songs[song_id]["recommendations"].append(recommendation)
            return True

        return False

    def delete_song(self, song_id):
        if song_id in self.songs:
            del self.songs[song_id]
            return True

        return False

    def save(self, filename="metadata.json"):
        with open(filename, "w") as file:
            json.dump(self.songs, file, indent=4)

    def load(self, filename="metadata.json"):
        if os.path.exists(filename):
            with open(filename, "r") as file:
                self.songs = json.load(file)

    def search_title(self, title):
        results = []

        for song in self.songs.values():
            if title.lower() in song["title"].lower():
                results.append(song)

        return results

    def search_artist(self, artist):
        results = []

        for song in self.songs.values():
            if artist.lower() in song["artist"].lower():
                results.append(song)

        return results
```
