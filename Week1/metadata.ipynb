```python
import uuid
```

```python
class SongMetadata:
    def __init__(self):
        self.songs = {}

    def add_song(self, title, artist, file_path, recommendations=None):
        for song_id, song in self.songs.items():
            if (
                song["title"].lower() == title.lower()
                and song["artist"].lower() == artist.lower()
            ):
                return song_id

        song_id = str(uuid.uuid4())

        self.songs[song_id] = {
            "title": title,
            "artist": artist,
            "file_path": file_path,
            "recommendations": recommendations if recommendations else []
        }

        return song_id

    def get_recommendations(self, song_id):
        if song_id in self.songs:
            return self.songs[song_id]["recommendations"]

        return []

    def add_recommendation(self, song_id, recommendation):
        if song_id in self.songs:
            self.songs[song_id]["recommendations"].append(recommendation)
            return True

        return False

    def search_title(self, title):
        results = []

        for song in self.songs.values():
            if title.lower() in song["title"].lower():
                results.append(song)

        return results

    def search_artist(self, artist):
        results = []

        for song in self.songs.values():
            if artist.lower() in song["artist"].lower():
                results.append(song)

        return results
```
