```python
import uuid
import json
import os
```

```python
class SongMetadata:
    def __init__(self):
        self.songs = {}

    def add_song(self, title, artist, file_path):
        for song_id, song in self.songs.items():
            if (
                song["title"].lower() == title.lower()
                and song["artist"].lower() == artist.lower()
            ):
                return song_id

        song_id = str(uuid.uuid4())

        self.songs[song_id] = {
            "title": title,
            "artist": artist,
            "file_path": file_path
        }

        return song_id

    def get_song(self, song_id):
        return self.songs.get(song_id)

    def get_all_songs(self):
        return self.songs
```
