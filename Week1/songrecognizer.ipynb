```python
#Aanya's code merged with Aahana's + Shriyan's folder
import uuid
import pickle #import to save and load files

class SongMetadata:
    def __init__(self):
        self.songs = {}

    def add_song(self, title, artist, file_path, recommendations=None):
        for song_id, song in self.songs.items():
            if (
                song["title"].lower() == title.lower()
                and song["artist"].lower() == artist.lower()
            ):
                return song_id

        song_id = str(uuid.uuid4())

        self.songs[song_id] = {
            "title": title,
            "artist": artist,
            "file_path": file_path,
            "recommendations": recommendations if recommendations else []
        }

        return song_id

    def get_recommendations(self, song_id):
        song = self.get_song(song_id)

        if song:
            return song["recommendations"]

        return []

    def add_recommendation(self, song_id, recommendation):
        if song_id in self.songs:
            self.songs[song_id]["recommendations"].append(recommendation)
            return True

        return False

    def search_artist(self, artist):
        results = []

        for song in self.songs.values():
            if artist.lower() in song["artist"].lower():
                results.append(song)

        return results

    #Aahana + Shriyan's code added below
    
    def get_song(self, song_id):#finds a specific song
        if song_id not in self.songs:
            print("ERROR: Song not found.")
            return None
        return self.songs.get(song_id)

    def get_all_songs(self): #all songs are returned
        return dict(self.songs)

    def num_songs(self): #total number of songs within the dictionary
        return len(self.songs)

    def contains(self, new_title, new_artist): #checks if song is in database
        target = (new_title.lower(), new_artist.lower())
        for song in self.songs.values():
            stored = (song["title"].lower(), song["artist"].lower())
            if stored == target:
                return True
        return False

    def delete_song(self, new_song_id): #deletes a song from the dictionary when ran
        if new_song_id not in self.songs:
            return self.songs
        return self.songs.pop(new_song_id)

    #save/load
    def save(self, path): #save the file in the system
        with open(path, mode="wb") as f:
            pickle.dump(self, f)

    @classmethod #makes it easier to call, less work 
    def load(cls, path): #loads the file 
        with open(path, "rb") as f:
            return pickle.load(f)
```
